```python
import requests
import json
from datetime import datetime

# 1. Configuración
# Usamos una API gratuita que no requiere registro para pruebas rápidas
URL_API = "https://api.exchangerate-api.com/v4/latest/USD"

# Ruta en tu Volumen de Databricks (ajustala a tu esquema)
PATH_VOLUMEN = "/Volumes/test/data_api_pib/data_api_volumen_1"

# Generamos el nombre con fecha para "guardarlo todo"
fecha_hoy = datetime.now().strftime("%Y-%m-%d")
nombre_archivo = f"tipo_cambio_{fecha_hoy}.json"
full_path = f"{PATH_VOLUMEN}/{nombre_archivo}"

# 2. Petición a la API
print(f"Consultando tipo de cambio para el {fecha_hoy}...")
response = requests.get(URL_API)

if response.status_code == 200:
    data = response.json()
    
    # 3. Guardar el JSON crudo en el Volumen
    with open(full_path, "w") as f:
        json.dump(data, f)
        
    print(f"¡Éxito! Archivo guardado en: {full_path}")
    print(f"Tipo de cambio USD/UYU hoy: {data['rates']['UYU']}")
else:
    print(f"Error al conectar: {response.status_code}")
```
