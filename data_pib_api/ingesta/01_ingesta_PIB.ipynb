```python
import requests
from datetime import datetime
import os

# 1. Parámetros
url = "https://api.worldbank.org/v2/country/all/indicator/NY.GDP.MKTP.CD?format=json&date=2000:2025&per_page=15000"

# 2. nombre de archivo único con la fecha y hora actual
timestamp = datetime.now().strftime("%Y%m%d_%H%M")
nombre_archivo = f"gdp_raw_{timestamp}.json"
path_base = "/Volumes/test/data_api_pib/data_api_volumen_1"
path_final = f"{path_base}/{nombre_archivo}"

# 3. Ejecución
print(f"Descargando datos en: {path_final}")
response = requests.get(url)

if response.status_code == 200:
    # Guardamos el contenido
    with open(path_final, "w") as f:
        f.write(response.text)
    print(f"¡Éxito! Archivo {nombre_archivo} guardado.")
else:
    print(f"Error al conectar con la API: {response.status_code}")
```
