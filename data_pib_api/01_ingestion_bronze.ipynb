```python
import requests

# URL del Banco Mundial (PIB de todos los países)
url = "https://api.worldbank.org/v2/country/all/indicator/NY.GDP.MKTP.CD?format=json&date=2000:2025&per_page=10000"
path_volumen = "/Volumes/test/data_api_pib/data_api_volumen_1/gdp.json"

response = requests.get(url)

if response.status_code == 200:
    with open(path_volumen, "w") as f:
        f.write(response.text)
    print("Archivo guardado en el Volumen!")
```
